```python
# ============================================================
# HOMEFEAST - SIMPLE GOOGLE COLAB VERSION
# ============================================================

!pip -q install flask

from flask import Flask, request, jsonify, render_template_string
from google.colab import output
import threading
import time

app = Flask(__name__)

# ============================================================
# SAMPLE FOOD DATA
# ============================================================

foods = [
    {
        "id": 1,
        "name": "South Indian Home Meal",
        "cook": "Lakshmi Home Foods",
        "type": "Veg",
        "cuisine": "South Indian",
        "price": 80,
        "plan": "Daily",
        "rating": 4.8,
        "time": "12:00 PM - 2:00 PM",
        "description": "Fresh rice, sambar, vegetables, rasam and curd."
    },
    {
        "id": 2,
        "name": "Healthy Veg Thali",
        "cook": "Annapoorna Kitchen",
        "type": "Veg",
        "cuisine": "North Indian",
        "price": 100,
        "plan": "Daily",
        "rating": 4.6,
        "time": "12:30 PM - 2:30 PM",
        "description": "Roti, dal, sabzi, rice and curd."
    },
    {
        "id": 3,
        "name": "Chicken Home Meal",
        "cook": "Muthu's Kitchen",
        "type": "Non-Veg",
        "cuisine": "South Indian",
        "price": 130,
        "plan": "Daily",
        "rating": 4.7,
        "time": "12:00 PM - 2:00 PM",
        "description": "Chicken curry, rice, vegetables and egg."
    },
    {
        "id": 4,
        "name": "Monthly Veg Tiffin",
        "cook": "Priya Home Meals",
        "type": "Veg",
        "cuisine": "South Indian",
        "price": 2200,
        "plan": "Monthly",
        "rating": 4.9,
        "time": "12:00 PM - 2:00 PM",
        "description": "Monthly homemade lunch subscription."
    },
    {
        "id": 5,
        "name": "Weekly Family Meal",
        "cook": "Home Taste Kitchen",
        "type": "Veg",
        "cuisine": "Indian",
        "price": 650,
        "plan": "Weekly",
        "rating": 4.5,
        "time": "12:00 PM - 2:00 PM",
        "description": "Affordable weekly home-style food."
    },
    {
        "id": 6,
        "name": "Fish Meals",
        "cook": "Meena's Kitchen",
        "type": "Non-Veg",
        "cuisine": "South Indian",
        "price": 150,
        "plan": "Daily",
        "rating": 4.8,
        "time": "12:30 PM - 2:30 PM",
        "description": "Fresh fish curry with rice and side dishes."
    }
]

orders = []

# ============================================================
# WEBSITE HTML
# ============================================================

HTML = """
<!DOCTYPE html>
<html>

<head>

<meta name="viewport" content="width=device-width, initial-scale=1">

<title>HomeFeast</title>

<style>

* {
    box-sizing: border-box;
    margin: 0;
    padding: 0;
}

body {
    font-family: Arial, sans-serif;
    background: #f7f7f7;
    color: #333;
}

/* NAVBAR */

.navbar {
    background: white;
    padding: 18px 7%;
    display: flex;
    justify-content: space-between;
    align-items: center;
    box-shadow: 0 2px 8px #ddd;
}

.logo {
    font-size: 28px;
    font-weight: bold;
    color: #e85d04;
}

.nav a {
    margin-left: 20px;
    text-decoration: none;
    color: #333;
}

/* HERO */

.hero {
    height: 430px;
    background:
    linear-gradient(rgba(0,0,0,.55),rgba(0,0,0,.55)),
    url("https://images.unsplash.com/photo-1547592180-85f173990554?auto=format&fit=crop&w=1500&q=80");

    background-size: cover;
    background-position: center;

    display: flex;
    justify-content: center;
    align-items: center;

    text-align: center;
    color: white;
}

.hero h1 {
    font-size: 45px;
    margin-bottom: 15px;
}

.hero p {
    font-size: 19px;
    margin-bottom: 25px;
}

.search {
    background: white;
    padding: 8px;
    border-radius: 8px;
    display: flex;
    max-width: 600px;
    margin: auto;
}

.search input {
    flex: 1;
    padding: 14px;
    border: none;
    outline: none;
    font-size: 16px;
}

.search button {
    background: #e85d04;
    color: white;
    border: none;
    padding: 0 25px;
    border-radius: 6px;
    cursor: pointer;
}

/* SECTION */

.section {
    padding: 50px 7%;
}

.title {
    text-align: center;
    margin-bottom: 30px;
}

.title h2 {
    font-size: 32px;
    margin-bottom: 8px;
}

/* FILTER */

.filters {
    background: white;
    padding: 20px;
    margin-bottom: 30px;
    border-radius: 10px;
}

.filters select {
    padding: 12px;
    margin-right: 10px;
    border: 1px solid #ddd;
    border-radius: 6px;
}

/* FOOD GRID */

.grid {
    display: grid;
    grid-template-columns: repeat(auto-fit,minmax(280px,1fr));
    gap: 25px;
}

.card {
    background: white;
    border-radius: 12px;
    overflow: hidden;
    box-shadow: 0 3px 12px #ddd;
}

.card img {
    width: 100%;
    height: 180px;
    object-fit: cover;
}

.card-content {
    padding: 20px;
}

.card h3 {
    margin-bottom: 8px;
}

.cook {
    color: #777;
}

.tags {
    margin: 12px 0;
}

.tag {
    display: inline-block;
    background: #fff0e5;
    color: #e85d04;
    padding: 5px 9px;
    border-radius: 20px;
    font-size: 12px;
    margin-right: 4px;
}

.description {
    color: #666;
    line-height: 1.5;
    margin-bottom: 15px;
}

.price-row {
    display: flex;
    justify-content: space-between;
    align-items: center;
}

.price {
    font-size: 21px;
    font-weight: bold;
    color: #e85d04;
}

button {
    cursor: pointer;
}

.order {
    background: #e85d04;
    color: white;
    border: none;
    padding: 10px 16px;
    border-radius: 6px;
}

/* FEATURES */

.features {
    background: white;
}

.feature-grid {
    display: grid;
    grid-template-columns: repeat(auto-fit,minmax(200px,1fr));
    gap: 20px;
}

.feature {
    text-align: center;
    padding: 25px;
}

.feature-icon {
    font-size: 40px;
    margin-bottom: 12px;
}

/* COOK */

.cook-section {
    background: #fff3e8;
    text-align: center;
}

.cook-section button {
    margin-top: 20px;
    padding: 13px 25px;
    border: none;
    background: #e85d04;
    color: white;
    border-radius: 6px;
}

/* FOOTER */

footer {
    background: #222;
    color: white;
    text-align: center;
    padding: 35px;
}

footer p {
    color: #bbb;
    margin-top: 10px;
}

/* MODAL */

.modal {
    display: none;
    position: fixed;
    left: 0;
    top: 0;
    width: 100%;
    height: 100%;
    background: rgba(0,0,0,.6);
    justify-content: center;
    align-items: center;
}

.modal-box {
    background: white;
    padding: 30px;
    width: 90%;
    max-width: 450px;
    border-radius: 10px;
}

.modal-box h2 {
    margin-bottom: 20px;
}

.modal-box input,
.modal-box select {
    width: 100%;
    padding: 12px;
    margin-bottom: 12px;
    border: 1px solid #ddd;
    border-radius: 6px;
}

.modal-box button {
    width: 100%;
    padding: 13px;
    background: #e85d04;
    color: white;
    border: none;
    border-radius: 6px;
}

.close {
    float: right;
    font-size: 25px;
    cursor: pointer;
}

/* MOBILE */

@media(max-width:700px) {

    .nav {
        display: none;
    }

    .hero h1 {
        font-size: 32px;
    }

}

</style>

</head>

<body>

<!-- NAVBAR -->

<div class="navbar">

    <div class="logo">
        🍱 HomeFeast
    </div>

    <div class="nav">

        <a href="#home">Home</a>

        <a href="#foods">Food</a>

        <a href="#how">How It Works</a>

        <a href="#cook">Become a Cook</a>

    </div>

</div>


<!-- HERO -->

<section class="hero" id="home">

<div>

<h1>Homemade Food, Made With Love ❤️</h1>

<p>
Fresh, healthy and affordable homemade meals
from local home cooks.
</p>

<div class="search">

<input
id="searchInput"
placeholder="Search meals or cooks..."
>

<button onclick="searchFood()">
Search
</button>

</div>

</div>

</section>


<!-- FOOD -->

<section class="section" id="foods">

<div class="title">

<h2>Popular Homemade Meals</h2>

<p>Choose your favourite homemade food</p>

</div>


<div class="filters">

<select id="typeFilter" onchange="filterFood()">

<option value="All">All Types</option>
<option value="Veg">Veg</option>
<option value="Non-Veg">Non-Veg</option>

</select>


<select id="planFilter" onchange="filterFood()">

<option value="All">All Plans</option>
<option value="Daily">Daily</option>
<option value="Weekly">Weekly</option>
<option value="Monthly">Monthly</option>

</select>


<select id="cuisineFilter" onchange="filterFood()">

<option value="All">All Cuisines</option>
<option value="South Indian">South Indian</option>
<option value="North Indian">North Indian</option>
<option value="Indian">Indian</option>

</select>

</div>


<div class="grid">

{% for food in foods %}

<div class="card food-item"

data-name="{{food.name}}"
data-cook="{{food.cook}}"
data-type="{{food.type}}"
data-plan="{{food.plan}}"
data-cuisine="{{food.cuisine}}">

<img src="https://images.unsplash.com/photo-1547592180-85f173990554?auto=format&fit=crop&w=800&q=80">

<div class="card-content">

<h3>{{food.name}}</h3>

<p class="cook">
👩‍🍳 {{food.cook}}
</p>

<div class="tags">

<span class="tag">
{{food.type}}
</span>

<span class="tag">
{{food.plan}}
</span>

<span class="tag">
⭐ {{food.rating}}
</span>

</div>

<p class="description">
{{food.description}}
</p>

<p>🕐 {{food.time}}</p>

<br>

<div class="price-row">

<div class="price">
₹{{food.price}}
</div>

<button class="order"
onclick="openOrder('{{food.name}}','{{food.price}}','{{food.plan}}')">

Subscribe

</button>

</div>

</div>

</div>

{% endfor %}

</div>

</section>


<!-- FEATURES -->

<section class="section features" id="how">

<div class="title">

<h2>Why Choose HomeFeast?</h2>

</div>

<div class="feature-grid">

<div class="feature">

<div class="feature-icon">🏠</div>

<h3>Homemade</h3>

<p>Fresh food prepared by local cooks.</p>

</div>


<div class="feature">

<div class="feature-icon">🥗</div>

<h3>Healthy</h3>

<p>Home-style and nutritious meals.</p>

</div>


<div class="feature">

<div class="feature-icon">💰</div>

<h3>Affordable</h3>

<p>Affordable daily meal plans.</p>

</div>


<div class="feature">

<div class="feature-icon">📅</div>

<h3>Subscriptions</h3>

<p>Daily, weekly and monthly plans.</p>

</div>

</div>

</section>


<!-- COOK -->

<section class="section cook-section" id="cook">

<h2>Are You a Home Cook? 👩‍🍳</h2>

<p>
Start your homemade food business with HomeFeast.
</p>

<button onclick="openCook()">
Become a Home Cook
</button>

</section>


<!-- FOOTER -->

<footer>

<h3>🍱 HomeFeast</h3>

<p>
Homemade food, made with love.
</p>

<p>
© 2026 HomeFeast
</p>

</footer>


<!-- ORDER MODAL -->

<div class="modal" id="orderModal">

<div class="modal-box">

<span class="close"
onclick="closeModal('orderModal')">
×
</span>

<h2>Subscribe to Meal</h2>

<h3 id="foodName"></h3>

<p id="foodPrice"></p>

<br>

<input
id="customerName"
placeholder="Your Name"
>

<input
id="customerPhone"
placeholder="Phone Number"
>

<input
id="customerAddress"
placeholder="Delivery Address"
>

<select id="customerPlan">

<option>Daily</option>
<option>Weekly</option>
<option>Monthly</option>

</select>

<button onclick="placeOrder()">
Confirm Subscription
</button>

</div>

</div>


<!-- COOK MODAL -->

<div class="modal" id="cookModal">

<div class="modal-box">

<span class="close"
onclick="closeModal('cookModal')">
×
</span>

<h2>Become a Home Cook</h2>

<input
id="cookName"
placeholder="Your Name"
>

<input
id="cookPhone"
placeholder="Phone Number"
>

<input
id="cookCuisine"
placeholder="Your Cuisine"
>

<button onclick="registerCook()">
Register
</button>

</div>

</div>


<script>

// SEARCH

function searchFood() {

    let search =
    document.getElementById("searchInput")
    .value.toLowerCase();

    let cards =
    document.querySelectorAll(".food-item");

    cards.forEach(card => {

        let name =
        card.dataset.name.toLowerCase();

        let cook =
        card.dataset.cook.toLowerCase();

        if (
            name.includes(search) ||
            cook.includes(search)
        ) {
            card.style.display = "block";
        }
        else {
            card.style.display = "none";
        }

    });

}


// FILTER

function filterFood() {

    let type =
    document.getElementById("typeFilter").value;

    let plan =
    document.getElementById("planFilter").value;

    let cuisine =
    document.getElementById("cuisineFilter").value;

    let cards =
    document.querySelectorAll(".food-item");

    cards.forEach(card => {

        let a =
        type === "All" ||
        card.dataset.type === type;

        let b =
        plan === "All" ||
        card.dataset.plan === plan;

        let c =
        cuisine === "All" ||
        card.dataset.cuisine === cuisine;

        if(a && b && c) {
            card.style.display = "block";
        }
        else {
            card.style.display = "none";
        }

    });

}


// ORDER

function openOrder(name, price, plan) {

    document.getElementById("foodName")
    .innerText = name;

    document.getElementById("foodPrice")
    .innerText = "Price: ₹" + price;

    document.getElementById("customerPlan")
    .value = plan;

    document.getElementById("orderModal")
    .style.display = "flex";

}


function placeOrder() {

    let name =
    document.getElementById("customerName").value;

    let phone =
    document.getElementById("customerPhone").value;

    let address =
    document.getElementById("customerAddress").value;

    let food =
    document.getElementById("foodName").innerText;

    let plan =
    document.getElementById("customerPlan").value;


    if(name === "" || phone === "" || address === "") {

        alert("Please fill all details.");

        return;
    }


    fetch("/order", {

        method: "POST",

        headers: {
            "Content-Type": "application/json"
        },

        body: JSON.stringify({

            name: name,
            phone: phone,
            address: address,
            food: food,
            plan: plan

        })

    })

    .then(response => response.json())

    .then(data => {

        alert(
            "🎉 Subscription Confirmed!\\n\\n" +
            "Food: " + food +
            "\\nPlan: " + plan
        );

        closeModal("orderModal");

    });

}


// COOK

function openCook() {

    document.getElementById("cookModal")
    .style.display = "flex";

}


function registerCook() {

    let name =
    document.getElementById("cookName").value;

    if(name === "") {

        alert("Please enter your name.");

        return;
    }

    alert(
        "Thank you " + name +
        "! Your registration has been submitted."
    );

    closeModal("cookModal");

}


// CLOSE

function closeModal(id) {

    document.getElementById(id)
    .style.display = "none";

}

</script>

</body>

</html>
"""


# ============================================================
# HOME PAGE
# ============================================================

@app.route("/")
def home():

    return render_template_string(
        HTML,
        foods=foods
    )


# ============================================================
# ORDER API
# ============================================================

@app.route("/order", methods=["POST"])
def order():

    data = request.json

    orders.append(data)

    print("NEW ORDER:", data)

    return jsonify({
        "success": True
    })


# ============================================================
# ORDERS
# ============================================================

@app.route("/orders")
def get_orders():

    return jsonify(orders)


# ============================================================
# START SERVER
# ============================================================

def start_server():

    app.run(
        host="0.0.0.0",
        port=5000,
        debug=False,
        use_reloader=False
    )


thread = threading.Thread(
    target=start_server
)

thread.daemon = True
thread.start()

time.sleep(3)


# ============================================================
# COLAB WEBSITE LINK
# ============================================================

url = output.eval_js(
    "google.colab.kernel.proxyPort(5000)"
)

print("\n" + "="*50)
print("       HOMEFEAST IS READY 🎉")
print("="*50)
print("\nOPEN YOUR WEBSITE HERE:\n")
print(url)
print("\n" + "="*50)
```
